# Basic Descriptive Analysis

This notebook contains only direct calls to the reusable descriptive-analysis class. Each code cell executes one analytical function and produces one output.


## Function: `analysis.datasets(include_excel=True)`

Lists every analytical dataset available in the repository, identifying the data layer (`raw` or `refined`), file format, and file size.


```python
from analysis_scripts import analysis
analysis.datasets(include_excel=True)
```

## Function: `analysis.overview_all("raw")`

Provides a structural overview of every raw dataframe: number of rows and columns, missing cells, duplicated rows, approximate memory use, and inferred temporal coverage.


```python
analysis.overview_all("raw")
```

## Function: `analysis.columns_all("raw")`

Inspects every column in every raw dataframe. The output identifies the dataset, column name, data type, non-null and missing counts, cardinality, and a concise interpretation of the field.


```python
analysis.columns_all("raw")
```

## Function: `analysis.overview("cumulative-answers-questions-stackexchange.csv")`

Shows the basic structure and quality information for the refined monthly dataset extracted from the legacy Excel workbook.


```python
analysis.overview("cumulative-answers-questions-stackexchange.csv")
```

## Function: `analysis.columns("cumulative-answers-questions-stackexchange.csv")`

## Function: `analysis.numeric_summary("cumulative-answers-questions-stackexchange.csv")`
